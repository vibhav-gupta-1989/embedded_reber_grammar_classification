# Generate valid embedded reber strings

```python
graph = [[['B', 1]], 
         [['T', 2], ['P', 3]],
         [['B', 4]],
         [['B', 11]],
         [['T', 5], ['P', 7]],
         [['S', 5], ['X', 6]],
         [['S', 9], ['X', 7]],
         [['T', 7], ['V', 8]],
         [['P', 6], ['V', 9]],
         [['E', 10]],
         [['T', 18]],
         [['T', 12], ['P', 14]],
         [['S', 12], ['X', 13]],
         [['S', 16], ['X', 14]],
         [['T', 14], ['V', 15]],
         [['P', 13], ['V', 16]],
         [['E', 17]],
         [['P', 18]],
         [['E', 19]]
         ]
```

```python
print(graph[18])
print(len(graph[0]))
```

```python
import random

s_list_positive = []
s_list_positive_labels = []
s_list_negative = []
s_list_negative_labels = []

letters = ['B', 'T', 'P', 'X', 'S', 'E', 'V']
count = 0
n = 10000

while (count < n):
    current_state = 0
    end_state = 19
    s = ""
    while(current_state != end_state):
        random_index = random.randint(0, len(graph[current_state])-1)
        s += graph[current_state][random_index][0]
        next_state = graph[current_state][random_index][1]
        current_state = next_state

    if(s not in s_list_positive):
        count = count + 1
        s_list_positive.append(s)
        s_list_positive_labels.append(0)

        s_negative = s
        random_index = random.randint(0, len(s)-1)
        random_index_letter = random.randint(0, len(letters)-1)

        while(letters[random_index_letter] == s[random_index]):
            random_index_letter = random.randint(0, len(letters)-1)

        s_negative = list(s_negative)
        s_negative[random_index] = letters[random_index_letter]
        s_negative = ''.join(s_negative)
        
        s_list_negative.append(s_negative)
        s_list_negative_labels.append(1)

print(len(s_list_negative))
print(len(s_list_positive))
   
    
```

# Create the train/validation/test datasets

```python
s_list_total = s_list_positive + s_list_negative

s_list_total_labels = s_list_positive_labels + s_list_negative_labels

combined = list(zip(s_list_total, s_list_total_labels))
random.shuffle(combined)

train_data = combined[:16000]
validation_data = combined[16000:18000]
test_data = combined[18000:]

print(train_data)
print(validation_data)
print(test_data)
```

# Create encoding for characters

```python
char_to_id = {
    '<PAD>': 0,
    'B': 1,
    'T': 2,
    'P': 3,
    'X': 4,
    'S': 5,
    'E': 6,
    'V': 7
}

def encode_string(s):
    return torch.tensor(
        [char_to_id[c] for c in s],
        dtype=torch.long
    )
```

# Create the loaders

```python
from torch.nn.utils.rnn import pad_sequence
from torch.utils.data import DataLoader
import torch

PAD_ID = 0

def collate_fn(batch):

    sequences, labels = zip(*batch)

    sequences = [encode_string(s) for s in sequences]

    sequences = pad_sequence(
        sequences,
        batch_first=True,
        padding_value=PAD_ID
    )

    labels = torch.tensor(labels, dtype=torch.long)

    return sequences, labels

train_loader = DataLoader(train_data, batch_size=32, collate_fn=collate_fn)
validation_loader = DataLoader(validation_data, batch_size=32, collate_fn=collate_fn)
test_loader = DataLoader(test_data, batch_size=32, collate_fn=collate_fn)
```

# Create the model

```python
import torch.nn as nn
class EmbeddedReberGrammar(nn.Module):
    def __init__(self, vocab_size, n_layers=2, embed_dim=128, hidden_dim=64, pad_id=0, dropout=0.2):
        super().__init__()
        self.embed = nn.Embedding(vocab_size, embed_dim, padding_idx=pad_id)
        self.gru = nn.GRU(embed_dim, hidden_dim, num_layers=n_layers, batch_first=True, dropout=dropout)
        self.output = nn.Linear(hidden_dim, 1)

    def forward(self, X):
        embeddings = self.embed(X)
        _outputs, hidden_states = self.gru(embeddings)
        return self.output(hidden_states[-1])

model = EmbeddedReberGrammar(8).to("cuda") 
```

# Define the training function

```python
import torchmetrics

def evaluate_tm(model, data_loader, metric):
    model.eval()
    metric.reset()
    with torch.no_grad():
        for X_batch, y_batch in data_loader:
            X_batch, y_batch = X_batch.to("cuda"), y_batch.to("cuda")
            y_pred = model(X_batch)
            metric.update(y_pred, y_batch.unsqueeze(1).float())
    return metric.compute()

def train(model, optimizer, loss_fn, metric, train_loader, valid_loader,
          n_epochs, patience=2, factor=0.5, epoch_callback=None):
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer, mode="max", patience=patience, factor=factor)
    history = {"train_losses": [], "train_metrics": [], "valid_metrics": []}
    for epoch in range(n_epochs):
        total_loss = 0.0
        metric.reset()
        model.train()
        if epoch_callback is not None:
            epoch_callback(model, epoch)
        for index, (X_batch, y_batch) in enumerate(train_loader):
            X_batch, y_batch = X_batch.to("cuda"), y_batch.to("cuda")
            y_pred = model(X_batch)
            loss = loss_fn(y_pred, y_batch.unsqueeze(1).float())
            total_loss += loss.item()
            loss.backward()
            optimizer.step()
            optimizer.zero_grad()
            metric.update(y_pred, y_batch.unsqueeze(1).float())
            train_metric = metric.compute().item()
            print(f"\rBatch {index + 1}/{len(train_loader)}", end="")
            print(f", loss={total_loss/(index+1):.4f}", end="")
            print(f", {train_metric=:.2%}", end="")
        history["train_losses"].append(total_loss / len(train_loader))
        history["train_metrics"].append(train_metric)
        val_metric = evaluate_tm(model, valid_loader, metric).item()
        history["valid_metrics"].append(val_metric)
        scheduler.step(val_metric)
        print(f"\rEpoch {epoch + 1}/{n_epochs},                      "
              f"train loss: {history['train_losses'][-1]:.4f}, "
              f"train metric: {history['train_metrics'][-1]:.2%}, "
              f"valid metric: {history['valid_metrics'][-1]:.2%}")
    return history
```

# Call the training function

```python
n_epochs = 20
xentropy = nn.BCEWithLogitsLoss()
optimizer = torch.optim.NAdam(model.parameters())
accuracy = torchmetrics.Accuracy(task="binary").to("cuda")
history = train(model, optimizer, xentropy, accuracy, train_loader, validation_loader, n_epochs)
```

# Evaluate on the test set

```python
evaluate_tm(model, test_loader, accuracy)
```
